```python
asia_cars = df[df["origin"] == "Asia"]
```

```python
selected = asia_cars[["vehicle_weight", "model_year"]]
```

```python
df["vehicle_weight"]                   # 一个Series
df[["vehicle_weight", "model_year"]]   # 一个DataFrame
```

```python
selected = selected.head(7)
selected
```

```python
X = selected.to_numpy()
X
```

```python
X.shape
```

```python
XTX = X.T @ X
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
y.shape
```

```python
w = XTX_inv @ X.T @ y
w
```
